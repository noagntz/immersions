## Exercice 1 – Produit de deux nombres

Dans cet exercice, on utilisera la fonction `int(input())` afin de demander des valeurs numériques à l’utilisateur. Par exemple,

```python
numéro = int(input("Choisissez un numéro")
```

À l'exécution, le code demandera à l'utilisateur de choisir un numéro.

On rappelle que l’instruction `if ... else` permet d’exécuter un bloc de code si une condition est vraie (`if`) et un autre bloc si elle est fausse (`else`). Par exemple,

```python
n = 5
if n > 0:
    print("Le nombre est positif")
else:
    print("Le nombre est négatif")
```

Si le nombre `n` est supérieur à 0, le programme doit afficher que le nombre est positif. Sinon, il doit afficher que le nombre est négatif.<br>
**Remarque : l’instruction `else` n’est pas obligatoire.**

### Question 1.1
Écrivez un code python permettant d'afficher le produit de deux nombres choisis par l'utilisateur. On appellera le premier nombre `a` et le second nombre `b`.<br>
On rappelle qu'un produit est une multiplication entre deux nombres.

Écrivez le code ci-dessous.

````python
# Définition des variables
a = int(input("Choisissez le premier nombre : "))
b = int(input("Choisissez le deuxième nombre : "))

# Calcul du produit
res = a * b

# Affichage du résultat
print(a, "x", b, "=", res)
````

### Question 1.2
À partir de votre code précédent, modifiez le programme de manière à ce que, si les deux nombres saisis sont égaux, le programme calcule et affiche le carré du nombre au lieu du produit.

Écrivez le code ci-dessous.

````python
# Définition des variables
a = int(input("Choisissez le premier nombre : "))
b = int(input("Choisissez le deuxième nombre : "))

# Vérification si les nombres sont égaux
if a == b:
    res = a**2
    print(a, "² =", res)
else:
    res = a * b
    print(a, "x", b, "=", res)
````

## Exercice 3 – Pair ou impair

Dans cet exercice, on utilisera la fonction `int(input())` et l'instruction `if ... else` comme dans l'exemple de l'*Exercice 1*, ainsi que l'opérateur modulo `%`.<br>
On rappelle que l’opérateur modulo `%` donne le reste d’une division entière. Par exemple,

```python
7 % 3 = 1
```
Le reste de la division entière 7÷3 est 1.<br>
**Remarque : Si le reste d'une division d'un nombre par 2 est 0, alors le nombre est pair.**

On rappelle également que si le reste d’un nombre divisé par 2 est 0, alors le nombre est pair.

### Question 3.1
Écrivez un code python permettant d'indiquer si le nombre choisis par l'utilisateur est pair ou impair. On appellera ce nombre `n`.

Écrivez le code ci-dessous.

````python
# Définition des variables
n = int(input("Entrez un nombre : "))

# Calcul pour savoir si le nombre est pair ou impair
if n % 2 == 0:
    print(n, "est pair")
else:
    print(n, "est impair")
````

## Exercice 6 – Devinette des couleurs

Dans cet exercice, on utilisera la fonction `int(input())` et l'instruction `if ... else` comme dans l'exemple de l'*Exercice 1*, ainsi qu'une boucle `while` comme dans l'exemple de l'*Exercice 2*.<br>
**Remarque : L'instruction `if ... else` peut contenir une ou plusieurs instructions intermédiaires `elif` pour tester d’autres conditions avant de passer au `else`.**

### Question 6.1
Complètez ce code python à trous permettant de créer le jeu de la devinette des couleurs, où le programme choisit une couleur au hasard dans la liste couleurs et l'utilisateur doit la trouver. À chaque essai, le programme doit indiquer si la couleur proposé est la bonne ou non.

Complètez le code ci-dessous.

````python
from random import choice

couleurs = ["rouge", "bleu", "vert", "jaune", "violet", "orange", "rose", "gris", "marron", "blanc", "noir"]

couleur_secrete = choice(couleurs)

nb_essai = 0
gagne = False

while not gagne:
    rep = input("Devinez la couleur choisie : \n ---> ").lower()
    
    if rep not in couleurs:
        print("Cette couleur n'est pas dans la liste. Essayez une autre.")
        continue

    if rep == couleur_secrete:
        print("Bravo ! Vous avez trouvé la couleur !")
        gagne = True
    else:
        print("Ce n'est pas la bonne couleur, essayez encore.")

    nb_essai += 1

print("Vous avez trouvé en", nb_essai, "essai(s)")
````

### Question 6.2
À partir de votre code précédent, modifiez le programme de manière à ce que cette fois, l'utilisateur ai une limite de réponse à pouvoir donner en plus de la verification de la couleur dans la liste. Si l'utilisateur à donné trop de mauvaise réponse, il a perdu.

Copiez-collez le code précédent, puis complètez le ci-dessous.

````python
from random import choice

couleurs = ["rouge", "bleu", "vert", "jaune", "violet", "orange", "rose", "gris", "marron", "blanc", "noir"]

couleur_secrete = choice(couleurs)

nb_essai = 0
gagne = False
max_essais = 5

while not gagne and nb_essai < max_essais:
    rep = input("Devinez la couleur choisie : \n ---> ").lower()
    
    if rep not in couleurs:
        print("Cette couleur n'est pas dans la liste. Essayez une autre.")
        continue

    if rep == couleur_secrete:
        print("Bravo ! Vous avez trouvé la couleur !")
        gagne = True
    else:
        print("Ce n'est pas la bonne couleur, essayez encore.")
    
    nb_essai += 1

if not gagne:
    print("Vous n'avez pas trouvé la couleur. C'était :", couleur_secrete)
else:
    print("Vous avez trouvé en", nb_essai, "essai(s)")
````

Dans cet exercice, on utilisera tout les concepts utilisé dans les différents exercices précédents, soit la fonction `int(input())` et l'instruction `if ... else` comme dans l'exemple de l'*Exercice 1*, une boucle `while` comme dans l'exemple de l'*Exercice 2*, l'opérateur modulo `%` comme dans l'exemple de l'*Exercice 3*, des boucles `for` comme dans l'exemple de l'*Exercice 4*, la méthode `append()` et la fonction `randint()` comme dans l'exemple de l'*Exercice 5*.

### Question 7.1
Complétez ce code python à trous permettant de créer le jeu du juste prix. Ce code permet de créer un jeu du juste prix dans lequel le programme choisit un nombre aléatoire entre 1 et 100, l’utilisateur doit le deviner en un nombre limité d’essais, et à chaque proposition le programme indique si le nombre à trouver est plus grand ou plus petit. Si l’utilisateur trouve le nombre, celui-ci est ajouté à une liste et son type (pair ou impair) est affiché, puis des statistiques sont calculées à la fin. On appellera ce nombre `n`.

Complètez le code ci-dessous.

````python
from random import randint

nombres_trouves = []

parties = int(input("Combien de parties voulez-vous jouer ? "))

for i in range(parties):
    nb_rand = randint(1, 100)
    nb_essai = 0
    gagne = False
    max_essais = 5

    print("\nNouvelle partie ! Devinez le nombre entre 1 et 100.")
    
    while not gagne and nb_essai < 5:
        rep_str = input("Quel est le juste prix ? \n ---> ")

        if not rep_str.isdigit():
            print("Ce n'est pas un nombre valide. Essayez à nouveau.")
            continue
        
        rep = int(rep_str)
        
        if rep > nb_rand:
            print("C'est moins, essayez encore !")
        elif rep < nb_rand:
            print("C'est plus, essayez encore !")
        else:
            print("C'est juste !")
            gagne = True
            nombres_trouves.append(nb_rand)
            if nb_rand % 2 == 0:
                print("Le nombre trouvé est pair.")
            else:
                print("Le nombre trouvé est impair.")
        
        nb_essai += 1

    if not gagne:
        print("Vous n'avez pas trouvé le nombre. C'était :", nb_rand)

pairs = 0
impairs = 0
somme = 0

for n in nombres_trouves:
    somme += n
    if n % 2 == 0:
        pairs += 1
    else:
        impairs += 1

print("\n=== Statistiques finales ===")
print("Nombres trouvés :", nombres_trouves)
print("Nombre de nombres pairs :", pairs)
print("Nombre de nombres impairs :", impairs)
print("Somme des nombres trouvés :", somme)
````
